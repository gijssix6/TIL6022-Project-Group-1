# 1. Introduction

The transition towards sustainable mobility requires a clear understanding of what drives travellers to choose public transport over private vehicles. While factors such as travel time, reliability, and comfort play an important role, the cost of travel remains one of the primary drivers behind commuting decisions. Over recent years, high volatility in fuel prices, combined with rising railway operational costs and fare adjustments, has continuously shifted the relative cost competitiveness between driving and taking the train.

This study investigates the impact of the ratio between fuel prices and train ticket prices on travellers' mode choice (car vs train). Rather than analysing the costs of a single mode, looking at the relative price ratio offers direct insight into the tipping point where passengers choose to change their mode of travel, without having to account for inflation.

# 3. Data Pipeline

To compare the relative price of rail travel to fuel costs, data from CBS will be used. Specifically, table 83131NED (*Consumentenprijzen; prijsindex 2015=100, 1996–2025*) is selected. This table provides the Consumer Price Index (CPI) across various expenditure categories, including passenger transport by train (subcategory 073110), diesel (072210), petrol (072220), and LPG (072230). By dividing the rail CPI by a weighted average fuel CPI, based on data from CLO, an annual fare-to-fuel price ratio is derived.

To determine the modal split, CBS table 84755NED (*Mobiliteitstrend; per rit, vervoerwijzen, reismotief, leeftijd en geslacht*) will be used. This dataset covers the period from 1999 to 2024 and contains the average number of trips made per person per transport mode, alongside the average trip distance. Multiplying these metrics yields the average daily kilometres travelled per person for each transport mode. Missing years within the time series will have to be interpolated.
